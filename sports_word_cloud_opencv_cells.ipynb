```python
# Import the libraries

import cv2
import numpy as np
from wordcloud import WordCloud
```

```python
# Sports-related words for the word cloud

sports_words = """
sports sports sports fitness fitness fitness cricket football
basketball football tennis badminton running swimming athletics
training training motivation motivation discipline discipline
teamwork teamwork health health strength stamina endurance
competition competition success success dedication practice
passion focus goal strategy energy skill achievement victory
sportsmanship wellness lifestyle determination performance
challenge game match league active balance inspiration
"""
```

```python
# Create the word cloud
# White background makes the design clean and easy to read.

wordcloud = WordCloud(
    width=1400,
    height=900,
    background_color="white",
    colormap="Dark2",
    prefer_horizontal=0.85,
    min_font_size=18,
    max_font_size=110,
    margin=8,
    random_state=42
).generate(sports_words)

cloud = np.array(wordcloud.to_image())
cloud = cv2.cvtColor(cloud, cv2.COLOR_RGB2BGR)

cv2.imshow("Sports Word Cloud", cloud)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Resize the image using OpenCV

cloud = cv2.resize(
    cloud,
    (1400, 900),
    interpolation=cv2.INTER_AREA
)
```

```python
# Add a white border so that the words do not touch the edges

cloud = cv2.copyMakeBorder(
    cloud,
    25, 25, 25, 25,
    cv2.BORDER_CONSTANT,
    value=(255, 255, 255)
)
```

```python
# Add my name in the centre of the image

name = "Aaradhya Pandey"
font = cv2.FONT_HERSHEY_SIMPLEX
font_scale = 2.2
thickness = 6

(text_width, text_height), baseline = cv2.getTextSize(
    name, font, font_scale, thickness
)

x = (cloud.shape[1] - text_width) // 2
y = (cloud.shape[0] + text_height) // 2

# White background behind the name for better readability
padding = 20
cv2.rectangle(
    cloud,
    (x - padding, y - text_height - padding),
    (x + text_width + padding, y + baseline + padding),
    (255, 255, 255),
    -1
)

cv2.putText(
    cloud,
    name,
    (x, y),
    font,
    font_scale,
    (128, 0, 128),
    thickness,
    cv2.LINE_AA
)
```

```python
# Add the main title

title = "SPORTS"
title_scale = 1.8
title_thickness = 5

(title_width, title_height), _ = cv2.getTextSize(
    title, font, title_scale, title_thickness
)

title_x = (cloud.shape[1] - title_width) // 2
title_y = y - text_height - 55

cv2.putText(
    cloud,
    title,
    (title_x, title_y),
    font,
    title_scale,
    (45, 70, 150),
    title_thickness,
    cv2.LINE_AA
)
```

```python
# Save the final image as a JPG

output_file = "sports_word_cloud.jpg"

cv2.imwrite(
    output_file,
    cloud,
    [cv2.IMWRITE_JPEG_QUALITY, 95]
)

print("Word cloud saved as:", output_file)
```

```python
# Display the final edited image

cv2.imshow("Sports Word Cloud - Aaradhya Pandey", cloud)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
